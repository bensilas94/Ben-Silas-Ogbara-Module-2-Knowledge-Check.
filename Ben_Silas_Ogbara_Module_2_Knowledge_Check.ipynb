Question Number 1.

Define “design envelope” in the context of subsea asset integrity. Why is the P-T envelope the most critical boundary for production systems?

Answer:

 Definition:

A design envelope is a multi-dimensional space defined by the safe operating limits of an asset across physical parameters such as pressure, temperature, flow rate, and environmental loads. Within the envelope, all parameters remain within certified limits; outside it, risk increases significantly.  

 Criticality of the P–T Envelope:

 The Pressure-Temperature (P–T) envelope is the most critical boundary because pressure dictates the mechanical stress and wall-thickness requirements (via Barlow's formula), while temperature dictates material yield strength (derating) and governs flow assurance risks such as hydrate formation and wax deposition.

 Calculate the minimum wall thickness using Barlow’s formula.

Internal Design Pressure ($P$) = 4,467 psi  

Outer Diameter ($D$) = 10.75 inches

Specified Minimum Yield Strength for X65 ($S$) = 65,000 psi

ASME B31.8 Design Factor ($F$) = 0.72

Longitudinal Joint Factor ($E$) = 1.0

Calculation Steps:Substitute the values into the formula:

$$t_{min} = \frac{4467 \times 10.75}{2 \times 65000 \times 0.72 \times 1.0}$$  

Calculate the numerator: $4467 \times 10.75 = 48020.25$

Calculate the denominator: $2 \times 65000 \times 0.72 \times 1.0 = 93600$

Perform the division: $t_{min} = \frac{48020.25}{93600} \approx 0.513$ inches

Answer: Minimum wall thickness $t_{min} = 0.513$ inches ($\approx 13.03$ mm).

A jumper has an original wall thickness of 12.7 mm, a current measured thickness of 10.8 mm after 10 years, and a code-minimum thickness of 8.5 mm. Calculate the average corrosion rate and the estimated remaining life. What inspection interval do you recommend?

Answer.

Average Corrosion Rate ($CR$):

$$\text{Total Wall Loss} = 12.7\text{ mm} - 10.8\text{ mm} = 1.9\text{ mm}$$

$$CR = \frac{1.9\text{ mm}}{10\text{ years}} = 0.19\text{ mm/year}$$

Estimated Remaining Useful Life ($RUL$):

$$\text{Allowable Wall Loss} = 10.8\text{ mm} - 8.5\text{ mm} = 2.3\text{ mm}$$

$$RUL = \frac{2.3\text{ mm}}{0.19\text{ mm/year}} \approx 12.1\text{ years}$$


Recommended Inspection Interval:

According to API 570 / NACE corrosion rate classifications, a corrosion rate of $0.19\text{ mm/yr}$ falls into the Medium Risk category ($0.13\text{ to } 0.25\text{ mm/yr}$) which dictates an inspection interval of 1 to 3 years. Given the proximity to the code minimum, an annual (1-year) inspection interval is recommended

Question Number 8.

What is Miner’s rule? A riser has accumulated damage ratios of 0.15 at 30 MPa, 0.08 at 50 MPa, and 0.03 at 80 MPa over 5 years. What is the total damage? What is the estimated total fatigue life?

Answers.

Miner’s Rule Definition:
Miner's rule estimates cumulative fatigue damage ($D$) by linearly adding the cycle damage ratios experienced across different stress amplitudes: $D = \sum \frac{n_i}{N_i}$, where failure occurs theoretically when $D = 1.0$.

Total Damage Calculation ($D_{total}$):$$D_{total} = 0.15 + 0.08 + 0.03 = 0.26 \text{ (over 5 years)}$$

Estimated Total Fatigue Life Calculation:$$\text
{Annual Damage Rate} = \frac{0.26}{5\text{ years}} = 0.052\text{ per year}$$   $$\text

{Total Fatigue Life} = \frac{1.0}{\text{Annual Damage Rate}} = \frac{1.0}{0.052} \approx 19.2\text{ years}$$

Question Number 9.

Explain why physics-informed ML models outperform pure data-driven approaches for asset integrity prediction. Give two specific examples

Answer:

Explanation:
Pure data-driven models rely entirely on historical correlations and often output physically impossible predictions when operating outside training bounds. Physics-informed ML integrates domain knowledge (laws of physics) to bound model behavior, requires less training data, and provides high interpretability and trust for safety-critical engineering decision.


Two Specific Examples:

1. Wall Thickness Prediction:
A physics-informed model incorporates Barlow's formula constraints to prevent negative wall thickness predictions, whereas a pure data-driven model might output mathematically valid but physically impossible negative values under extreme error distributions.

2. Fatigue Life Forecasting: Incorporating Miner's rule and S–N curve principles restricts cumulative damage accumulation rates logically, avoiding erratic predictions typical of unconstrained black-box regressions.
